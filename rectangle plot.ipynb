```python
import cv2
```

```python
image=cv2.imread("vk.png")
```

```python
if image is None:
    print("image is not working ")
else:
    pt1=(50,500)
    pt2=(400,300)
    color=(255,0,0)
    thickness=4
    cv2.rectangle(image,pt1,pt2,color,thickness)
    cv2.imshow("line", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
